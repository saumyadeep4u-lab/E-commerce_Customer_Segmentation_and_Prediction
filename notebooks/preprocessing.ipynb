- Duplicate Handling

```python
# Imports
import pandas as pd
import numpy as np
```

```python
# Load data
DATA_PATH = r"C:\Users\hp\Downloads\BIA\E-commerce_Customer_Segmentation_and_Prediction\data\data.csv"
data = pd.read_csv(DATA_PATH, encoding='latin1') # Using latin1 to handle non-UTF-8 characters
data.head()
```

```python
# Data back-up
df = data.copy()
df.info()
```

```python
# Meta data
print("Number of rows: ", df.shape[0])
print("Number of columns: ", df.shape[1])
all_columns = df.columns.to_list()
num_columns = df.select_dtypes(include = "number").columns.to_list()
cat_columns = df.select_dtypes(include = "object").columns.to_list()
print("\nAll columns: \n", all_columns)
print("\nNumeric columns: \n", num_columns)
print("\nCategorical columns: \n", cat_columns)
print("\nData type: \n", df.dtypes)
```

```python
# Check missing values
print("Missing values in each column:")
print(df.isnull().sum())
```

```python
#Standardizing Product Descriptions Using Majority Value per Stock Code
df['Description'] = (df.groupby('StockCode')['Description'].transform(lambda x: x.mode()[0] if not x.mode().empty else x.iloc[0]))
```

```python
df[df['Description'].isna()] # Filters and displays all rows where the Description column has missing (NaN) values
```

```python
df['Description'] = df['Description'].fillna('Unknown') # Filling null values with Unknown
```

```python
df['Description'].isna().sum() #Checking
```

```python
# Duplicate
print("Duplicate details: ", df.duplicated().sum())
print("Duplicate rows: \n")
df[df.duplicated()]
```

```python
df[df.duplicated(keep=False)] # Display all exact duplicate rows (including first occurrences)
```

```python
df = df.drop_duplicates() # Remove all exact duplicate rows from the dataset
df.duplicated().sum() # Verify that no duplicate rows remain
```

```python
# Converting InvoiceDate to datetime format for feature extraction
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])

# Extracting time-based features from InvoiceDate
# These features help in analyzing sales trends and customer behavior over time
df['Year'] = df['InvoiceDate'].dt.year
df['Month'] = df['InvoiceDate'].dt.month
df['Day'] = df['InvoiceDate'].dt.day
df['Hour'] = df['InvoiceDate'].dt.hour
df['DayOfWeek'] = df['InvoiceDate'].dt.day_name()

# Converting identifier and categorical columns to 'category' data type
# to improve memory efficiency and optimize performance for grouping and analysis
df['InvoiceNo'] = df['InvoiceNo'].astype('str')
df['StockCode'] = df['StockCode'].astype('category')
df['CustomerID'] = pd.to_numeric(df['CustomerID'], errors='coerce')
df['Country'] = df['Country'].astype('category')
```

```python
print(df.isnull().sum())
```

```python
df['CustomerID'].isna().sum()  # Count missing values
```

```python
# Create Total Price 
df['TotalPrice'] = df['Quantity'] * df['UnitPrice']  # Includes both purchases (+) and returns (-)
```

```python
# Create Return Feature
df['Return'] = df['Quantity'] < 0  # Flag return transactions
```

```python
# Split Known vs Unknown Customers
df_known = df[df['CustomerID'].notna()].copy()   # For RFM & clustering
df_unknown = df[df['CustomerID'].isna()].copy()  # For analysis only
```

```python
# Save Cleaned Data for RFM,CLV and Recommendation Models
KNOWN_DATA_PATH = r"C:\Users\hp\Downloads\BIA\E-commerce_Customer_Segmentation_and_Prediction\data\known.csv"
df_known.to_csv(KNOWN_DATA_PATH, index = False) 

UNKNOWN_DATA_PATH = r"C:\Users\hp\Downloads\BIA\E-commerce_Customer_Segmentation_and_Prediction\data\unknown.csv"
df_unknown.to_csv(UNKNOWN_DATA_PATH, index = False)
```
